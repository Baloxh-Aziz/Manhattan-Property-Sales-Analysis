```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_excel("rollingsales_manhattan.xls", skiprows=4)
```

```python
avg_price = data.groupby('NEIGHBORHOOD')['SALE PRICE'].mean()

# Sort values
avg_price = avg_price.sort_values(ascending=False)
avg_price.head(10).plot(kind='bar')

plt.title("Top 10 Neighborhoods by Average Sale Price")
plt.xlabel("Neighborhood")
plt.ylabel("Average Price")
plt.xticks(rotation=45)

plt.show()
```

#Insight
The bar chart shows the top 10 neighborhoods based on average sale price.
It is clear that some neighborhoods have much higher prices than others.
This shows that property prices are not the same in all areas.
Location has a strong impact on property value, as some areas are more expensive and in higher demand.

```python
data["SALE DATE"] = pd.to_datetime(data["SALE DATE"])

# group by month
data["MONTH"] = data["SALE DATE"].dt.to_period("M")
monthly_avg = data.groupby("MONTH")["SALE PRICE"].mean()
monthly_avg.plot(kind="line")

plt.title("Average Sale Price Over Time")
plt.ylabel("Sale Price")
plt.show()
```

#Insight
The line graph shows how the average sale price changes over time (month by month).
It can be observed that property prices do not remain constant and fluctuate over different months.
Some months show higher average prices while others show lower values.
This indicates that there may be seasonal trends or market changes affecting property prices over time.

```python
# Extract month
data['MONTH'] = data['SALE DATE'].dt.month

monthly_sales = data.groupby('MONTH')['SALE PRICE'].mean()

monthly_sales.plot()

plt.title("Average Sale Price by Month (2012–2013)")
plt.xlabel("Month")
plt.ylabel("Average Price")

plt.show()
```

#Insight
The graph shows that average sale prices are different in each month. 
Some months have higher prices while others have lower prices. 
This means property prices change over time and are affected by market demand.

```python
# Boxplot to detect outliers
plt.boxplot(data['SALE PRICE'])
plt.title("Sale Price Outliers")
plt.show()
```

#Insight
Boxplot shows that the Sale Price distribution contains outliers,
indicating that some properties have significantly higher or 
lower prices compared to the majority of the data.
This may affect analysis and may require further investigation or handling.
